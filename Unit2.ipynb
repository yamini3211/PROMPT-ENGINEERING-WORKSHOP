```python
from groq import Groq
from google.colab import userdata
client = Groq(
    api_key=userdata.get("apikey3")
)
```

```python
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": "Hello World"}
    ]
)

print(response.choices[0].message.content)
```

```python
#UNIT-2:EXPERIMENT-5
#Comparison of zero-shot and few-shot prompting
```

```python
zero_prompt = """
Classify the sentiment of the following review as
Positive, Negative, or Neutral.

Review:
"The laptop performance is excellent but the battery drains quickly."

Answer only with the sentiment.
"""

zero_response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": zero_prompt}
    ]
)

print(zero_response.choices[0].message.content)
```

```python
few_prompt = """
You are a sentiment classifier.

Example 1
Review: "I love this mobile."
Sentiment: Positive

Example 2
Review: "Worst customer service ever."
Sentiment: Negative

Example 3
Review: "The product is okay."
Sentiment: Neutral

Now classify:

Review:
"The laptop performance is excellent but the battery drains quickly."

Sentiment:
"""

few_response =  client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "user", "content": few_prompt}
    ]
)


print(few_response.choices[0].message.content)
```
